```python
# The tokenizer creates the model inputs, input_ids and attention_mask. The model’s forward method only accepts input_ids and attention_mask, so 
# set remove_columns to drop columns like horoscope after tokenization.

# Set truncation=True and a max_length to truncate longer sequences to a specified maximum length.
# Use the train_test_split method to create a test split for evaluating the model.
from datasets import load_dataset
from transformers import AutoTokenizer, DataCollatorForLanguageModeling

model_name = "Qwen/Qwen3-0.6B"
tokenizer = AutoTokenizer.from_pretrained(model_name)
dataset = load_dataset("karthiksagarn/astro_horoscope", split="train")

def tokenize(batch):
    return tokenizer(
        batch["horoscope"],
        truncation=True,
        max_length=512,
    )

dataset = dataset.map(tokenize, batched=True, remove_columns=dataset.column_names)
dataset = dataset.train_test_split(test_size=0.1)
```
